RAG Example with streamlit

```python
#1. Import the libraries

import os
from dotenv import load_dotenv
# For loading the PDF document
from langchain_community.document_loaders import PyPDFLoader
# For splitting the text into chunks
from langchain_text_splitters import RecursiveCharacterTextSplitter
# For creating embeddings
from langchain_openai import OpenAIEmbeddings
# For vector store
from langchain_community.vectorstores import FAISS
```

```python
#2. Load API key
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
#3. Load the PDF document
pdf_loader = PyPDFLoader("data/EMPLOYEE_AGREEMENT.pdf")
pdf_pages = pdf_loader.load()
```

```python
#4. Split the text into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
doc_chunks = text_splitter.create_documents([page.page_content for page in pdf_pages])
print("Total Chunks:", len(doc_chunks))
```

```python
#5. Create embeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
#doc_embeddings = embeddings.embed_documents([chunk.page_content for chunk in doc_chunks])
```

```python
#6. Create vector store and save to disk
vector_store = FAISS.from_documents(doc_chunks, embeddings)
vector_store.save_local("vectorstore")
print("Vector store created and saved successfully.")
```
