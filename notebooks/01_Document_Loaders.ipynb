```python
#Openai GPT model
from google.colab import userdata
import os

os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")

from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-4o-mini")

result = model.invoke("What is the capital of India?")
print(result.content)
```

```python
result = model.invoke("Who is Donal Trump. Expalin in 30 words")
print(result)
```

```python
#Google Gemini key
from langchain_google_genai import ChatGoogleGenerativeAI
import os

os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

response = llm.invoke("Explain Virat Kohli in 30 words.")
print(response.content)
```

```python
result = llm.invoke("How many leaves I can take in  my company?")
print(result.content)
```

```python
# Txt data loader

from langchain_community.document_loaders import TextLoader

# Create loader
loader = TextLoader("cricket.txt")

# Load the document
documents = loader.load()

# Print text content
print(documents[0].page_content)

# Print metadata
print(documents[0].metadata)
```

```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://en.wikipedia.org/wiki/Cosine_similarity")

docs = loader.load()

print(docs[0].page_content)
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-4o-mini")

question = "What is cosine similarity ?"

response = model.invoke(f"""
Answer the question using only the content below.

Content:
{docs[0].page_content}

Question:
{question}
""")

print(response.content)
```
