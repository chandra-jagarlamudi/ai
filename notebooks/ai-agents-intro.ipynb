```python
# OPEN API call example using langchain-openai

from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os

load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

model = ChatOpenAI(model="gpt-4o-mini")

result = model.invoke("What is the capital of India?")
print(result.content)
```

```python
# Google Gemini call example using langchain-googlegemini

from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

response = llm.invoke("Explain Sathya Sai in 30 words.")
print(response.content)

```

```python
# Txt data loader

from langchain_community.document_loaders import TextLoader

# Create loader
loader = TextLoader("cricket.txt")

# Load the document
documents = loader.load()

# Print text content
print(documents[0].page_content)

# Print metadata
print(documents[0].metadata)
```

```python
# Web data loader

from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://en.wikipedia.org/wiki/Cosine_similarity")

docs = loader.load()

print(docs[0].page_content)
```
